```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../outputs/revenue_analysis.csv")
```

```python
df.head()
```

```python
print(df["price"].sum())
```

```python
top_products = df.groupby("product_id")["price"] \
    .sum() \
    .sort_values(ascending=False) \
    .head(10)

print(top_products)
```

```python
top_products.plot(kind="bar")

plt.title("Top 10 Products by Revenue")
plt.xlabel("Product ID")
plt.ylabel("Revenue")

plt.show()
```

```python
print(df["price"].mean())
```

```python
summary = pd.DataFrame({
    "Metric": ["Total Revenue", "Average Price"],
    "Value": [
        df["price"].sum(),
        df["price"].mean()
    ]
})

summary.to_csv("../outputs/business_summary.csv", index=False)
```
